# The sigmoid function,
- The sigmoid function also known as the logistic function, is a mathematical function commonly used in machine learning, particularly in neural networks as an activation function. It transforms an input value into a range between 0 and 1, producing an S-shaped curve. 

# Gradient descent 
- Gradient descent is an optimization algorithm which is commonly-used to train machine learning models and neural networks. It trains machine learning models by minimizing errors between predicted and actual results.

```python
import numpy as np
```

```python
# Sigmoid Function
def sigmoid(z):
    return 1.0/(1.0 + np.exp(-z))


# Calculating Gradieant Decent

def cal_gradieant(theta , X,y): 
    m = y.size # number of samples
    # (X @ theta) -> linear combination
    # sigmoid(X @ theta) -> predictions
    # (sigmoid - y) -> error
    # X.T @ (...) / m -> average gradient for all samples
    return (X.T @ (sigmoid(X @ theta) - y))/ m


# Gradieant Decent    
def gradieant_decent(X,y,alpha= 0.1,num_iter = 100, tol = 1e-7):
    X_b  = np.c_[np.ones((X.shape[0],1)),X] # add bias column
    theta  = np.zeros(X_b.shape[1])         # initialize weights = 0
    for i in range (num_iter):
        grad = cal_gradieant(theta,X_b,y)   # compute gradient
        theta -= alpha*grad                 # update weights
        if np.linalg.norm(grad) < tol:      # stop if updates are tiny
            break
    return theta


# Predict probabilities (between 0 and 1)   
def predict_proba(X,theta):
    X_b = np.c_[np.ones((X.shape[0],1)),X]   # add bias term
    return sigmoid(X_b @ theta)


# Predict binary class (0 or 1) using threshold (default = 0.5) 
def predict(X,theta,thresold=0.5):
    return (predict_proba(X,theta)>=thresold).astype(int)  


# Accuracy metric: fraction of correct predictions
def accuracy(actual,predicted):
    actual = np.array(actual)
    predicted = np.array(predicted)
    
    correct = np.sum(actual == predicted)
    total = len(actual)
    return correct / total


 # X = input features.
 # y = labels (0 or 1).
 # alpha = learning rate.S

 
```

```python
import pandas as pd 
# Data for heart_disease
df = pd.read_csv(r"C:\Users\lakha\Downloads\heart_disease_health_indicators_BRFSS2015.csv\heart_disease_health_indicators_BRFSS2015.csv")
```

```python
df.columns
```

```python
X = df[['HighBP', 'HighChol', 'CholCheck', 'BMI',
       'Smoker', 'Stroke', 'Diabetes', 'PhysActivity', 'Fruits', 'Veggies',
       'HvyAlcoholConsump', 'AnyHealthcare', 'NoDocbcCost', 'GenHlth',
       'MentHlth', 'PhysHlth', 'DiffWalk', 'Sex', 'Age', 'Education',
       'Income']]
y = df['HeartDiseaseorAttack']
```

```python
# Train And Split the data

np.random.seed(13)  # for reproducibility
indices = np.arange(len(X))
np.random.shuffle(indices)
X = X.iloc[indices]
y = y.iloc[indices]
split_idx = int(len(X) * 0.8) # 80 % data for train

# Split The data
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

# Now Scaling the X
train_mean = X_train.mean()
train_std = X_train.std()
# we scaled the test data and used train mean becuse in reality model dont have large test data so it does not have own mean and sometime it can give bias prediction
X_train_scaled = (X_train - train_mean) / train_std
X_test_scaled  = (X_test - train_mean) / train_std

# Updating the theta Value
theta_hat = gradieant_decent(X_train_scaled, y_train,alpha= 0.1)

# Now predicting the values
y_pred_train = predict(X_train_scaled,theta_hat)
y_pred_test = predict(X_test_scaled,theta_hat)
```

```python

print(accuracy(y_train,y_pred_train))
print(accuracy(y_test,y_pred_test))
```
